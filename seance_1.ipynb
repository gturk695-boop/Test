### A.4 Les opérations

```python
# Arithmétique sur les nombes

#On utilise print() pour que tout s'affiche

a, b = 17, 5
print("a + b  =", a + b)
print("a - b  =", a - b)
print("a * b  =", a * b)

print("a / b  =", a / b)     # division classique
print("a // b =", a // b)    # division entière (quotient)
print("a % b  =", a % b)     # reste de la division (modulo)
print("a ** b =", a ** b)    # puissance
```

### A.6 Les boucles

```python
liste_taux = [1.0956, 1.0919, 1.0940, 1.0895, 1.0921]

# Boucle for : on parcourt chaque élément de la liste
somme = 0
for t in liste_taux:
    print(t)
    somme = somme + t
print("somme avec for :", round(somme, 4))
print("moyenne :", round(somme / len(liste_taux), 4))
```

```python
# Boucle while : on répète tant que la condition est vraie, avec un compteur
somme = 0
i = 0
while i < len(liste_taux):
    somme = somme + liste_taux[i]
    i = i + 1
print("somme avec while :", round(somme, 4))

# Vérification avec la fonction intégrée sum
print("somme avec sum() :", round(sum(liste_taux), 4))
```

```python
# on ne sait pas à l'avance combien de tours il faut

# QUESTION : Combien d'années pour doubler un capital placé à 5 % par an ?
capital = 100
annees = 0
while capital < 200:
    capital = capital * 1.05
    annees = annees + 1
print(annees, "ans, capital =", round(capital, 2))
```

### C.1 Explorer un dictionnaire

```python
reponse = {
    "amount": 1.0,
    "base": "EUR",
    "date": "2024-01-02",
    "rates": {"USD": 1.0956},
}

# 1. Accès aux valeurs
print(reponse["date"])
print(reponse["rates"])
print(reponse["rates"]["USD"])     # dictionnaire imbriqué : deux crochets
```

```python
# 2. Affecter une valeur à une clé qui existe déjà
reponse["date"] = "2099-01-01"
print(reponse)
print("nombre de clés :", len(reponse))   # toujours 4, la clé n'est pas dupliquée
```

La valeur est remplacée
Il n'y a pas de deuxième clé `"date"`

```python
# 3a) copie avec .copy() puis modification d'une clé simple
reponse = {"amount": 1.0, "base": "EUR", "date": "2024-01-02", "rates": {"USD": 1.0956}}

copie = reponse.copy()
copie["date"] = "2099-01-01"
print("copie    :", copie)
print("original :", reponse)       # l'original n'a pas bougé
```

```python
# 3b) modif° de la partie imbriquée rates dans la copie
copie["rates"]["USD"] = 9.99
print("copie    :", copie)
print("original :", reponse)       # l'original est modifié aussi !

# Les deux dictionnaires pointent vers le même dictionnaire rates
print(copie["rates"] is reponse["rates"])
```

```python
# 4) Copie profonde avec le module copy
import copy

reponse = {"amount": 1.0, "base": "EUR", "date": "2024-01-02", "rates": {"USD": 1.0956}}

copie_profonde = copy.deepcopy(reponse)
copie_profonde["rates"]["USD"] = 9.99
print("copie profonde :", copie_profonde)
print("original       :", reponse)      # cette fois l'original est protégé
print(copie_profonde["rates"] is reponse["rates"])
```

### C.2 Appeler l'API

```python
import json
import csv
import urllib.request
from pathlib import Path

DEVISE = "USD"      
BASE = "EUR"
```

**1. Taux du jour**

```python
url = f"https://api.frankfurter.app/latest?from={BASE}&to={DEVISE}"
requete = urllib.request.Request(url, headers={"User-Agent": "Mozilla/5.0"})

with urllib.request.urlopen(requete, timeout=10) as r:
    texte = r.read().decode("utf-8")   # la réponse arrive en texte (JSON)

donnees = json.loads(texte)              # texte JSON -> dictionnaire Python
print(type(donnees))
print(donnees)
print(f"Date : {donnees['date']}, 1 {BASE} = {donnees['rates'][DEVISE]} {DEVISE}")
```

**2. Série du 1er janvier au 1er septembre 2026**

On regarde d'abord si la réponse est déjà dans `cache/`, et si oui on la relit. sinon on appelle l'API et on enregistre la réponse brute.

```python
debut, fin = "2026-01-01", "2026-09-01"

dossier_cache = Path("cache")
dossier_cache.mkdir(exist_ok=True)
fichier_cache = dossier_cache / f"taux_{BASE}_{DEVISE}_{debut}_{fin}.json"

if fichier_cache.exists():
    print("Lecture du cache :", fichier_cache)
    serie = json.loads(fichier_cache.read_text(encoding="utf-8"))
else:
    url = f"https://api.frankfurter.app/{debut}..{fin}?from={BASE}&to={DEVISE}"
    print("Appel de l'API :", url)
    requete = urllib.request.Request(url, headers={"User-Agent": "Mozilla/5.0"})
    with urllib.request.urlopen(requete, timeout=10) as r:
        texte = r.read().decode("utf-8")
    fichier_cache.write_text(texte, encoding="utf-8")   # on garde la réponse brute
    serie = json.loads(texte)

print(serie.keys())
print("du", serie["start_date"], "au", serie["end_date"])
print("nombre de jours :", len(serie["rates"]))
```

==>> La partie `rates` est maintenant un dictionnaire indexé par date

```python
dates = list(serie["rates"].keys())
print(dates[:3])
print(serie["rates"][dates[0]])
```

On écrit une version simple en CSV 

```python
dossier_donnees = Path("donnees")
dossier_donnees.mkdir(exist_ok=True)
fichier_csv = dossier_donnees / f"taux_{BASE}_{DEVISE}.csv"

with open(fichier_csv, "w", newline="", encoding="utf-8") as f:
    ecrivain = csv.writer(f)
    ecrivain.writerow(["date", "taux"])               # ligne d'en-tête
    for date in sorted(serie["rates"]):               # dates triées dans l'ordre
        ecrivain.writerow([date, serie["rates"][date][DEVISE]])

print("CSV écrit :", fichier_csv)
```

Vérification

```python
with open(fichier_csv, encoding="utf-8") as f:
    lignes = f.readlines()

print("nombre de lignes (avec l'en-tête) :", len(lignes))
for ligne in lignes[:6]:
    print(ligne.strip())
```
